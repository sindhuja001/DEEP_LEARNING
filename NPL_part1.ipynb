```python
import nltk
from nltk import word_tokenize

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

a = "Sindhu is working in Google in UK"

# Tokenization
b = word_tokenize(a)

# Part of Speech Tagging
c = nltk.pos_tag(b)

# Named Entity Recognition
d = nltk.ne_chunk(c)

# Display output
print(b)
print(c)
print(d)
```

```python
#sentiment analysis
from sklearn.feature_extraction.text import CountVectorizer

feedbacks=[
    "delivery  is good good",
    "delivery is late",
    "good customer service",
    "delivery is worst",
    "package is worst"
      ]
bow=CountVectorizer()
b1=bow.fit_transform(feedbacks)

for i in feedbacks:
  print(i)

print(bow.get_feature_names_out())
print(b1.toarray())
```

```python
#Tf-idf
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
d=[

   "python course for beginners",
   "python programming lang",
   "laptop delivery and customer service"
]

tf = TfidfVectorizer()
abc=tf.fit_transform(d)

abcd=pd.DataFrame(
    abc.toarray(),
    columns=tf.get_feature_names_out()

)

print(abcd.round(2))
```
